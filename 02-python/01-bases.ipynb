# Python de base — pour programmeurs C/C++

Ce notebook couvre les fondements de Python en partant du point de vue d'un programmeur C/C++.  
Chaque concept est mis en parallèle avec son équivalent C/C++ pour accélérer la compréhension.

## 1. Variables et types

### En C/C++
```cpp
int x = 10;
double pi = 3.14;
bool flag = true;
std::string name = "Benoit";
```
Le type est déclaré explicitement et lié à la variable pour toute sa durée de vie.

### En Python
Le type est lié à **l'objet**, pas à la variable. La variable n'est qu'une **référence** vers un objet en mémoire — exactement comme un pointeur, mais géré automatiquement.

```python
x = 10        # x pointe vers un objet int
x = "Benoit"  # x pointe maintenant vers un objet str — complètement valide
```

C'est le **typage dynamique** : le type est déterminé à l'exécution, pas à la compilation.

````python
x = 10
print(f"x = {x}, type = {type(x)}")

pi = 3.14
print(f"pi = {pi}, type = {type(pi)}")

flag = True
print(f"flag = {flag}, type = {type(flag)}")

name = "Benoit"
print(f"name = {name}, type = {type(name)}")
````

### Tout est un objet

En C++, `int` est un type primitif. En Python, **même un entier est une instance de classe**.  
La fonction `type()` retourne la classe de l'objet — l'équivalent de `typeid()` en C++.

On peut vérifier l'identifiant unique de l'objet en mémoire avec `id()` — analogue à l'adresse d'un pointeur :

````python
x = 10
print(f"id de x : {id(x)}")  # adresse mémoire de l'objet

y = x  # y pointe vers le MÊME objet que x
print(f"id de y : {id(y)}")  # même adresse

y = 20  # y pointe maintenant vers un NOUVEL objet
print(f"id de y après y=20 : {id(y)}")  # adresse différente
print(f"x est toujours : {x}")  # x n'a pas changé
print(f"id de x : {id(x)}")  # adresse mémoire de l'objet
````

## 2. Structures de contrôle

### En C/C++
```cpp
if (x > 0) {
    printf("positif\n");
} else {
    printf("négatif\n");
}
```

### En Python
**Pas d'accolades.** L'indentation définit les blocs — c'est une contrainte syntaxique, pas une convention.  
**Pas de parenthèses** autour de la condition (mais elles sont tolérées).

````python
x = 10

if x > 0:
    print("positif")
elif x == 0:
    print("zéro")
else:
    print("négatif")
````

### Boucles

**`while`** — identique à C/C++ mais sans parenthèses ni accolades :

````python
i = 0
while i < 3:
    print(f"i = {i}")
    i += 1
````

**`for`** — très différent du C/C++. En Python, `for` itère toujours sur une **séquence**, jamais sur un compteur.

```cpp
// C++ : boucle sur un compteur
for (int i = 0; i < 5; i++) { ... }
```

```python
# Python : itération sur une séquence
for i in range(5): ...
```

`range(5)` génère la séquence `[0, 1, 2, 3, 4]` — c'est un objet itérable, pas un simple entier.

````python
# Équivalent de : for (int i = 0; i < 5; i++)
for i in range(5):
    print(i)

print("---")

# Équivalent de : for (int i = 2; i < 8; i += 2)
for i in range(2, 8, 2):  # range(début, fin, pas)
    print(i)
````

## 3. Fonctions

### En C/C++
```cpp
int addition(int a, int b) {
    return a + b;
}
```

### En Python
- Mot-clé `def` à la place du type de retour
- Pas de types déclarés (mais on peut les annoter — on verra ça plus tard)
- Indentation à la place des accolades

````python
def addition(a, b):
    return a + b

print(addition(3, 4))
print(addition("Hello ", "World"))  # fonctionne aussi avec des strings !
````

### Valeurs par défaut et arguments nommés

Python supporte les **arguments avec valeur par défaut** (comme C++ depuis C++11)  
et les **arguments nommés** — une fonctionnalité absente de C.

````python
def saluer(nom, greeting="Bonjour"):
    print(f"{greeting}, {nom}!")

saluer("Benoit")                    # utilise la valeur par défaut
saluer("Benoit", "Salut")          # remplace la valeur par défaut
saluer(greeting="Hey", nom="Benoit")  # arguments nommés, ordre libre
````

## 4. Structures de données principales

### Liste — l'équivalent du tableau dynamique (`std::vector`)

- Taille variable
- Peut contenir des types mixtes (contrairement à `std::vector<T>`)
- Indexation à partir de 0 (comme C/C++)

````python
ma_liste = [10, 20, 30, 40, 50]

print(ma_liste[0])   # premier élément : 10
print(ma_liste[-1])  # dernier élément : 50 (index négatif = depuis la fin)
print(ma_liste[1:3]) # slicing : éléments index 1 et 2 → [20, 30]

ma_liste.append(60)  # équivalent de push_back()
print(ma_liste)
````

### Tuple — liste immuable

Comme une liste, mais **non modifiable après création** — analogue à un tableau `const` en C++.  
Utile pour des données qui ne doivent pas changer (coordonnées, configuration, etc.).

````python
coordonnees = (48.8566, 2.3522)  # latitude, longitude Paris

print(coordonnees[0])  # lecture OK

# coordonnees[0] = 0  # → TypeError : un tuple est immuable
````

### Dictionnaire — l'équivalent de `std::map` ou d'une struct

Paires clé→valeur. Les clés peuvent être n'importe quel type immuable (string, int, tuple...).

````python
personne = {
    "nom": "Benoit",
    "age": 40,
    "langages": ["C", "C++", "LabVIEW", "Python"]
}

print(personne["nom"])       # accès par clé
print(personne["langages"])  # la valeur peut être une liste

personne["ville"] = "Montréal"  # ajout d'une nouvelle clé
print(personne)
````

## 5. Strings et f-strings

### Différences avec C/C++

| C/C++ | Python |
|---|---|
| `char *` ou `std::string` | `str` |
| Terminateur `\0` | Pas de terminateur (objet avec longueur stockée) |
| `strlen(s)` | `len(s)` |
| `printf("%d items", n)` | `f"{n} items"` |
| Concaténation : `strcat` ou `+` (avec `std::string`) | `+` ou f-string |

Une `str` Python est **immuable** — comme un tuple. Toute "modification" crée une nouvelle string.

### Les f-strings (Python 3.6+)

L'équivalent moderne et idiomatique du `printf` : tu mets l'expression directement dans la chaîne, préfixée par `f`. L'évaluation a lieu au moment de la création de la string.

````python
prenom = "Benoit"
age = 40

# Style ancien (à éviter en code moderne)
print("Bonjour " + prenom + ", tu as " + str(age) + " ans")
print("Bonjour %s, tu as %d ans" % (prenom, age))
print("Bonjour {}, tu as {} ans".format(prenom, age))

# Style moderne — f-string
print(f"Bonjour {prenom}, tu as {age} ans")

# f-string avec formatage
pi = 3.14159265
print(f"pi arrondi à 2 décimales       : {pi:.2f}")
print(f"pi en notation scientifique     : {pi:.3e}")
print(f"pi sur 10 caractères à droite   : {pi:>10.3f}")

# Méthodes courantes
texte = "  Hello Python  "
print(texte.strip())                    # supprime les espaces aux extrémités
print(texte.upper())                    # majuscules
print(texte.replace("Python", "World"))
print("a,b,c,d".split(","))             # → ['a', 'b', 'c', 'd']
print("-".join(["a", "b", "c"]))        # → 'a-b-c'
````

## 6. Set — collection sans doublons

L'équivalent de `std::unordered_set` (table de hachage). Stocke des éléments **uniques**, accès en O(1) en moyenne.

**Cas d'usage typique :** test d'appartenance rapide, suppression de doublons, opérations ensemblistes (union, intersection, différence).

Le set est **mutable** — comme `list` et `dict`. Pour une version immuable, utilise `frozenset`.

````python
# Créer un set
fruits = {"pomme", "banane", "cerise"}
print(fruits)

# Ajouter / retirer
fruits.add("kiwi")
fruits.discard("banane")
print(fruits)

# Test d'appartenance (O(1) en moyenne — bien plus rapide qu'une liste)
print("pomme" in fruits)

# Supprimer les doublons d'une liste
liste_avec_doublons = [1, 2, 2, 3, 3, 3, 4]
liste_unique = list(set(liste_avec_doublons))
print(liste_unique)

# Opérations ensemblistes
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(f"union              : {a | b}")
print(f"intersection       : {a & b}")
print(f"différence (a-b)   : {a - b}")
print(f"symétrique (xor)   : {a ^ b}")
````

## 7. Mutabilité et passage par référence — LE piège pour un dev C

C'est ici que les habitudes C/C++ deviennent dangereuses. En Python :

- Les variables ne stockent **jamais** de valeurs, elles stockent des **références vers des objets** (cf. section 1, `id()`).
- Quand tu passes une variable à une fonction, tu passes la **référence**, pas une copie.
- Mais certains objets sont **immuables** (int, float, str, tuple) et d'autres **mutables** (list, dict, set).

### Tableau des mutabilités

| Type | Mutable ? | Conséquence |
|---|---|---|
| `int`, `float`, `bool` | Non | Toute modif crée un nouvel objet |
| `str` | Non | `s += "x"` crée une nouvelle string |
| `tuple` | Non | Lecture seule |
| `list` | Oui | `.append()`, `[i] = ...` modifient en place |
| `dict` | Oui | Ajout/suppression de clés modifient en place |
| `set` | Oui | `.add()` modifie en place |

### Le piège #1 — modification d'un argument mutable

Une fonction qui reçoit une liste peut la modifier — et l'appelant verra la modification. C'est l'équivalent en C de passer un `std::vector<int>&` (référence non-const). Pas un piège en soi, mais une surprise quand on s'attend au comportement valeur de C.

````python
def ajouter_zero(liste):
    liste.append(0)  # modifie l'objet pointé, pas la référence locale

ma_liste = [1, 2, 3]
ajouter_zero(ma_liste)
print(ma_liste)  # → [1, 2, 3, 0] — l'appelant voit la modification !

# Comparé à un int (immuable)
def doubler(n):
    n = n * 2  # crée un nouvel objet, réassigne la variable locale uniquement
    return n

x = 10
doubler(x)
print(x)  # → 10 — x n'a pas bougé
````

### Le piège #2 — valeur par défaut mutable

LE piège classique de Python, sur lequel tout débutant se fait avoir. Une valeur par défaut est créée **une seule fois**, à la définition de la fonction. Si c'est un objet mutable, il est **partagé entre tous les appels** qui n'en fournissent pas une.

````python
def ajouter_item(item, liste=[]):  # ⚠️ DANGER : liste par défaut mutable
    liste.append(item)
    return liste

print(ajouter_item("a"))   # → ['a']
print(ajouter_item("b"))   # → ['a', 'b']  ← surprise : la liste est PARTAGÉE

# Le bon pattern : sentinelle None
def ajouter_item_safe(item, liste=None):
    if liste is None:
        liste = []  # nouvelle liste à CHAQUE appel
    liste.append(item)
    return liste

print(ajouter_item_safe("a"))  # → ['a']
print(ajouter_item_safe("b"))  # → ['b']  ← OK
````

## 8. `is` vs `==` — identité vs égalité

| Opérateur | Compare | Équivalent C++ |
|---|---|---|
| `==` | **valeur** (via `__eq__`) | `operator==` surchargé |
| `is` | **identité** (même objet en mémoire) | `&a == &b` |

`is` revient à comparer deux `id()`. Deux objets différents peuvent avoir la même valeur (`==` vrai, `is` faux).

**Règle pratique :**
- `is` uniquement pour comparer avec `None`, `True`, `False` : `if x is None:`
- `==` partout ailleurs

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(f"a == b : {a == b}")   # True  — mêmes valeurs
print(f"a is b : {a is b}")   # False — objets différents en mémoire
print(f"a is c : {a is c}")   # True  — c pointe vers le même objet que a

# Cas particulier : CPython cache les petits ints (-5 à 256)
x = 100
y = 100
print(f"100 is 100 : {x is y}")     # True — cache d'entiers (optimisation CPython)

x = 1000
y = 1000
print(f"1000 is 1000 : {x is y}")   # peut être False — au-delà du cache

# Le cas pratique : tester None
def traiter(valeur=None):
    if valeur is None:        # idiomatique
        print("valeur absente")
    else:
        print(f"valeur : {valeur}")

traiter()
traiter(42)
````

## 9. Vérité (truthiness) — différent du C

En C, `if (x)` revient à `if (x != 0)`. En Python c'est plus riche : **n'importe quel objet** peut être évalué comme booléen via la fonction interne `bool(obj)`.

### Valeurs falsy (évaluées à False)

- `False`
- `None`
- Zéro numérique : `0`, `0.0`, `0j`
- Conteneurs vides : `""`, `[]`, `()`, `{}`, `set()`
- Tout objet dont la classe définit `__bool__()` retournant `False`

### Tout le reste est truthy

Y compris `"False"` (chaîne non vide), `[0]` (liste non vide), `-1` (entier non nul).

````python
# Le pattern pythonique pour tester "non vide"
ma_liste = [1, 2, 3]
if ma_liste:                # équivalent à `if len(ma_liste) > 0:`, mais idiomatique
    print("La liste a des éléments")

ma_liste = []
if not ma_liste:
    print("La liste est vide")

# Vérifier la valeur bool() de divers objets
print(f"bool('')         : {bool('')}")        # False
print(f"bool('False')    : {bool('False')}")   # True — chaîne non vide
print(f"bool([])         : {bool([])}")        # False
print(f"bool([0])        : {bool([0])}")       # True — liste non vide
print(f"bool(0)          : {bool(0)}")         # False
print(f"bool(-1)         : {bool(-1)}")        # True
print(f"bool(None)       : {bool(None)}")      # False
````

## 10. Compréhensions de listes — pythonique et essentiel pour numpy/pandas

C'est une syntaxe condensée pour construire une liste à partir d'une autre séquence. **Critique** parce qu'elle revient partout dans pandas, numpy, et le code Python idiomatique.

### Forme générale

```python
[expression for item in iterable if condition]
```

### Équivalent C/C++

```cpp
std::vector<int> result;
for (int i : source) {
    if (condition(i)) {
        result.push_back(transform(i));
    }
}
```

### Équivalent Python "naïf" (style C porté en Python)

```python
result = []
for i in source:
    if condition(i):
        result.append(transform(i))
```

### Équivalent Python idiomatique (compréhension)

```python
result = [transform(i) for i in source if condition(i)]
```

Pourquoi c'est mieux : plus court, lu comme une expression mathématique (set-builder notation), et souvent plus rapide à l'exécution (boucle optimisée en C dans CPython).

````python
# Carrés des nombres de 0 à 9
carres = [x**2 for x in range(10)]
print(carres)

# Avec filtrage : carrés des nombres pairs uniquement
carres_pairs = [x**2 for x in range(10) if x % 2 == 0]
print(carres_pairs)

# Avec une condition ternaire dans l'expression
signes = ["+" if x > 0 else "-" if x < 0 else "0" for x in [-2, -1, 0, 1, 2]]
print(signes)

# Compréhension de dict
mots = ["pomme", "banane", "cerise"]
longueurs = {mot: len(mot) for mot in mots}
print(longueurs)

# Compréhension de set (élimine les doublons)
caracteres_uniques = {c for c in "hello world" if c != " "}
print(caracteres_uniques)

# Imbriqué : matrice 3x3
matrice = [[i*3 + j for j in range(3)] for i in range(3)]
for ligne in matrice:
    print(ligne)
````

## 11. `enumerate` et `zip` — itérer comme un pro

### L'anti-pattern C-style

```python
# Mauvais (style C porté en Python)
for i in range(len(ma_liste)):
    print(i, ma_liste[i])
```

C'est verbeux, sujet aux erreurs off-by-one, et ne respecte pas l'esprit Python.

### `enumerate` — index + valeur en même temps

Retourne des tuples `(index, valeur)`. Equivalent moral d'une boucle C++ avec un compteur séparé.

### `zip` — itérer en parallèle sur plusieurs séquences

Quand tu as deux (ou plus) listes "alignées" et veux les parcourir en même temps. S'arrête à la plus courte.

````python
# enumerate
fruits = ["pomme", "banane", "cerise"]
for i, fruit in enumerate(fruits):
    print(f"{i}: {fruit}")

print("---")

# enumerate avec un start différent
for i, fruit in enumerate(fruits, start=1):
    print(f"#{i}: {fruit}")

print("---")

# zip — itérer en parallèle
prenoms = ["Alice", "Bob", "Charlie"]
ages = [30, 25, 35]
for prenom, age in zip(prenoms, ages):
    print(f"{prenom} a {age} ans")

print("---")

# zip s'arrête à la séquence la plus courte
longue = [1, 2, 3, 4, 5]
courte = ["a", "b", "c"]
print(list(zip(longue, courte)))  # → [(1,'a'), (2,'b'), (3,'c')]

# zip + enumerate combinés (cas réel)
for i, (prenom, age) in enumerate(zip(prenoms, ages)):
    print(f"{i}: {prenom}, {age} ans")
````

## 12. Annotations de types — la promesse tenue

La section 3 mentionnait *"pas de types déclarés (mais on peut les annoter)"*. Voici comment.

Depuis Python 3.5, on peut **annoter** les types des paramètres et du retour. Ces annotations sont **purement documentaires à l'exécution** — Python ne vérifie rien. Mais :

- Les IDE (VSCode, PyCharm) les utilisent pour l'auto-complétion et les warnings inline
- Les outils statiques comme `mypy`, `pyright` les vérifient au build — équivalent moral du compilateur C++ qui vérifie les types

### Forme

```python
def fonction(param: Type) -> TypeRetour:
    ...
```

C'est l'équivalent moral des signatures C++ typées, mais **sans contrôle compilateur par défaut**. Le typage reste dynamique au runtime — les annotations sont "optionnelles et facultatives".

````python
def addition(a: int, b: int) -> int:
    return a + b

# Python n'empêche pas ça à l'exécution :
print(addition(3, 4))           # OK : 7
print(addition("a", "b"))       # OK : "ab" — Python s'en fiche (mypy gueulerait)

# Annotations pour structures complexes
from typing import Optional

def trouver_utilisateur(id: int) -> Optional[dict]:
    """Retourne le user dict ou None si introuvable."""
    base = {1: {"nom": "Alice"}, 2: {"nom": "Bob"}}
    return base.get(id)

print(trouver_utilisateur(1))   # {'nom': 'Alice'}
print(trouver_utilisateur(99))  # None

# Variables aussi peuvent être annotées
nom: str = "Benoit"
age: int = 40
coordonnees: tuple[float, float] = (48.85, 2.35)

# Lire les annotations d'une fonction au runtime
print(addition.__annotations__)
````

## 13. Exceptions — `try`/`except`

L'équivalent du `try`/`catch` du C++. Différences principales :

- Plus utilisé en Python qu'en C++ — beaucoup d'opérations *normales* lèvent des exceptions (clé absente, fichier inexistant, type incompatible)
- La philosophie pythonique est **EAFP** ("Easier to Ask Forgiveness than Permission") : essayer puis attraper, plutôt que vérifier avant
- Toutes les exceptions héritent de `Exception` (qui hérite de `BaseException`)

### Hiérarchie utile à connaître

```
BaseException
 └── Exception
      ├── ValueError      (mauvaise valeur d'un type correct)
      ├── TypeError       (mauvais type)
      ├── KeyError        (clé absente dans un dict)
      ├── IndexError      (index hors limites)
      ├── FileNotFoundError
      └── ... etc.
```

````python
# Cas typique : conversion qui peut échouer
def convertir_en_int(s):
    try:
        return int(s)
    except ValueError as e:
        print(f"Erreur de conversion : {e}")
        return None

print(convertir_en_int("42"))     # → 42
print(convertir_en_int("hello"))  # → None (avec message d'erreur)

# Plusieurs types d'exceptions + finally
def lire_dict(d, cle):
    try:
        return d[cle]
    except KeyError:
        print(f"Clé '{cle}' absente")
        return None
    except TypeError:
        print("d n'est pas un dict")
        return None
    finally:
        print("Fin de la tentative.")

print(lire_dict({"a": 1}, "a"))   # → 1
print(lire_dict({"a": 1}, "z"))   # → None (KeyError attrapée)

# Lever une exception manuellement
def diviser(a, b):
    if b == 0:
        raise ValueError("Division par zéro")
    return a / b

try:
    diviser(10, 0)
except ValueError as e:
    print(f"Capturé : {e}")
````

## 14. Modules et imports

Un **module** = un fichier `.py`. Un **package** = un dossier avec un fichier `__init__.py` (souvent vide, juste pour marquer le dossier comme package).

### Quatre façons d'importer

```python
import math                       # accès via math.sqrt(...)
import math as m                  # accès via m.sqrt(...)
from math import sqrt             # accès direct via sqrt(...)
from math import sqrt as racine   # alias direct
```

### Bonnes pratiques

- **`import xxx`** — préférable pour clarifier d'où vient une fonction (`math.sqrt` plutôt que juste `sqrt`)
- **`from xxx import yyy`** — pour des noms ultra-fréquents ou conventionnels (`from typing import Optional`)
- **`from xxx import *`** — à éviter en production (pollue le namespace, casse l'auto-complétion)

### Mécanisme d'import

Quand Python rencontre `import math`, il :

1. Cherche `math` dans `sys.path` (liste de dossiers à inspecter)
2. Trouve `math.py` (ou un binaire `.so` Linux / `.pyd` Windows pour les modules natifs)
3. Exécute le fichier **une seule fois**
4. Stocke le module dans `sys.modules` — les imports suivants utilisent ce cache

C'est l'équivalent moral du `#include` du C, sauf que :
- C'est exécuté **au runtime**, pas par le préprocesseur
- Il n'y a **pas de header séparé** — le fichier `.py` contient à la fois "déclaration" et "implémentation"
- Le résultat est mis en cache → pas de double inclusion possible (alors qu'en C il faut les `#pragma once` ou les include guards)

````python
import math

print(math.sqrt(16))
print(math.pi)
print(math.factorial(5))

# import avec alias
import math as m
print(m.cos(0))    # 1.0

# from ... import — pour les noms qu'on utilise souvent
from math import sqrt, pi
print(sqrt(25))
print(pi)

# Voir où Python cherche les modules (les dossiers de sys.path)
import sys
print("\nDossiers de recherche (sys.path) :")
for p in sys.path[:5]:
    print(f"  {p}")
print(f"  ... ({len(sys.path)} entrées au total)")

# Voir les modules déjà importés (et donc en cache)
print(f"\nNombre de modules en cache : {len(sys.modules)}")
print(f"math est-il en cache ? {'math' in sys.modules}")
````
